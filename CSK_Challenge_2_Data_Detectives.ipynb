```python
import pandas as pd
```

```python
data = {
    "hours": [1, 2, 3, 4, 5, 2, 4, 6],
    "scores": [35, 42, 58, 70, 85, 48, 73, 91]
}

df = pd.DataFrame(data)

print(df)
```

```python
df.shape
```

```python
df.columns
```

```python
df["scores"]
```

```python
average_score = df["scores"].mean()
print(average_score)
```

```python
average_score = df["scores"].min()
print(average_score)
```

```python
average_score = df["scores"].max()
print(average_score)
```
